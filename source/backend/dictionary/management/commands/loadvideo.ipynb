```python
import pandas as pd

df = pd.read_csv('seed_data.csv')
video_links = df['video_link']
print(video_links)
```

```python
from selenium import webdriver
import time

# Khởi tạo trình duyệt (ví dụ dùng Chrome)
driver = webdriver.Chrome()

for link in reversed(video_links):
    driver.get(link)
    time.sleep(2)  # Đợi 2 giây để trang tải xong

driver.quit()

```

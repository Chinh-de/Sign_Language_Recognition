```python
import pandas as pd
import numpy as np
```

```python
df_glossID_to_index = pd.read_csv('gloss_id_to_index.csv')
top_300 = pd.read_csv('top_300_classes.txt', sep='\t', header=None, names=['gloss_id', 'gloss'])
print(top_300.head())
df_glossID_to_index = df_glossID_to_index.merge(top_300[['gloss_id', 'gloss']], on='gloss_id', how='left')
print(df_glossID_to_index.head())
```

```python
df_glossID_to_index.to_csv('index_to_gloss.csv', index=False, columns=['index', 'gloss'])
```

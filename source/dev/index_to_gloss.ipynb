```python
import pandas as pd
import numpy as np
```

```python
df_glossID_to_index = pd.read_csv('gloss_id_to_index.csv')
top_300 = pd.read_csv('top_300_classes.txt', sep='\t', header=None, names=['gloss_id', 'gloss'])
print(top_300.head())
df_glossID_to_index = df_glossID_to_index.merge(top_300[['gloss_id', 'gloss']], on='gloss_id', how='left')
print(df_glossID_to_index.head())
```

```python
df_glossID_to_index.to_csv('index_to_gloss.csv', index=False, columns=['index', 'gloss'])
```

```python
df_video_gloss_mapping = pd.read_csv('video_gloss_mapping.csv')
# file 100 từ sử dụng cho model
df_glossID_to_index = pd.read_csv('gloss_id_to_index.csv')
# Giữ lại những row có gloss_id nằm trong df_glossID_to_index
df_video_gloss_mapping = df_video_gloss_mapping[df_video_gloss_mapping['gloss_id'].isin(df_glossID_to_index['gloss_id'])]

df_subset = pd.read_csv('all_gloss_video_subset.csv')


df_video_gloss_mapping = df_video_gloss_mapping.merge(
    df_subset[['video_id', 'subset']],
    on='video_id',
    how='left'
)


top_300_df = pd.read_csv('top_300_classes.txt', sep='\t', header=None, names=['gloss_id', 'gloss'])
top_300_df.to_csv('top_300_classes.csv', index=False)

df_video_gloss_mapping = df_video_gloss_mapping.merge(df_glossID_to_index, on='gloss_id', how='left')
df_video_gloss_mapping = df_video_gloss_mapping.merge(top_300_df, left_on='gloss_id', right_on='gloss_id', how='left')
df_video_gloss_mapping['gloss'] = df_video_gloss_mapping['gloss'].fillna('')
df_video_gloss_mapping.head(100)

df_video_gloss_mapping.to_csv('seed_data.csv', index=False, columns=['video_id', 'gloss_id', 'gloss', 'video_link', 'subset'])
```
